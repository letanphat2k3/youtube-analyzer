```python
from fastapi import FastAPI
from fastapi.middleware.cors import CORSMiddleware
from pydantic import BaseModel
from transformers import Qwen2VLForConditionalGeneration, AutoProcessor
from qwen_vl_utils import process_vision_info
from yt_dlp import YoutubeDL
import torch
import uvicorn
import nest_asyncio
from pyngrok import ngrok
from google.colab import userdata
```

```python
ngrok_token = userdata.get('NGROK_TOKEN')
if not ngrok_token:
    raise Exception("❌ Bạn chưa lưu NGROK_TOKEN! Dùng: userdata.set('NGROK_TOKEN', 'your_token')")

ngrok.set_auth_token(ngrok_token)
```

```python
app = FastAPI()
```

```python
app.add_middleware(
    CORSMiddleware,
    allow_origins=["*"],
    allow_credentials=True,
    allow_methods=["*"],
    allow_headers=["*"],
)
```

```python
MODEL_NAME = "Qwen/Qwen2-VL-7B-Instruct"

model = Qwen2VLForConditionalGeneration.from_pretrained(
    MODEL_NAME,
    torch_dtype=torch.bfloat16,
    device_map="auto",
    attn_implementation="flash_attention_2",
)

processor = AutoProcessor.from_pretrained(MODEL_NAME)
```

```python
class YoutubeRequest(BaseModel):
    url: str

class AskVideoRequest(BaseModel):
    question: str
```

```python
from fastapi.responses import FileResponse
import os

@app.get("/favicon.ico")
def favicon():
    favicon_path = "favicon.ico"
    if os.path.exists(favicon_path):
        return FileResponse(favicon_path)
    return {"detail": "No favicon available"}
```

```python
@app.post("/download_video")
async def download_video(req: YoutubeRequest):
    ydl_opts = {
        "format": "best",
        "outtmpl": "m.mp4"
    }
    try:
        with YoutubeDL(ydl_opts) as ydl:
            ydl.download([req.url])
        return {"message": "✅ Video đã được tải về thành công."}
    except Exception as e:
        return {"error": str(e)}
```

```python
latest_result = {"status": "chưa xử lý", "question": None, "answer": None}
```

```python
@app.post("/analyze_video")
async def analyze_video(data: AskVideoRequest):
    global latest_result
    try:
        messages = [
            {
                "role": "user",
                "content": [
                    {"type": "video", "video": "m.mp4", "max_pixels": 1280 * 780, "fps": 0.1},
                    {"type": "text", "text": data.question},
                ],
            }
        ]
        text = processor.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
        image_inputs, video_inputs = process_vision_info(messages)
        inputs = processor(
            text=[text],
            images=image_inputs,
            videos=video_inputs,
            padding=True,
            return_tensors="pt"
        ).to("cuda")

        generated_ids = model.generate(**inputs, max_new_tokens=512)
        generated_ids_trimmed = [out_ids[len(in_ids):] for in_ids, out_ids in zip(inputs.input_ids, generated_ids)]
        output_text = processor.batch_decode(
            generated_ids_trimmed,
            skip_special_tokens=True,
            clean_up_tokenization_spaces=False
        )

        latest_result = {
            "status": "đã xử lý",
            "question": data.question,
            "answer": output_text[0]
        }

        return {"response": output_text[0]}
    except Exception as e:
        latest_result = {
            "status": "lỗi",
            "question": data.question,
            "answer": str(e)
        }
        return {"error": str(e)}
```

```python
@app.get("/result")
def get_latest_result():
    return latest_result
```

```python
public_url = ngrok.connect(8000)
print("🔗 Ngrok public URL:", public_url)

nest_asyncio.apply()
uvicorn.run(app, host="0.0.0.0", port=8000)
```
